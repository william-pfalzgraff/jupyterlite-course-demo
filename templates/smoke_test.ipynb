```python
import os, sys, platform
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
print('Python', platform.python_version(), '| numpy', np.__version__, '| matplotlib', matplotlib.__version__)
```

```python
# 1. Is file storage active? (work is only saved if this is true)
cwd = os.getcwd()
print(('✓' if cwd.startswith('/drive') else '✗'), 'working folder is', cwd)

# 2. Can a file be written and read back?
try:
    with open('_site_check.txt', 'w') as f: f.write('ok')
    with open('_site_check.txt') as f: ok = f.read() == 'ok'
    os.remove('_site_check.txt')
    print(('✓' if ok else '✗'), 'file write and read')
except Exception as e:
    print('✗', 'file write and read:', e)

# 3. Numerical libraries
try:
    import scipy; print('✓', 'scipy', scipy.__version__)
except Exception as e:
    print('✗', 'scipy:', e)
```

```python
# 4. An interactive figure: it should have a toolbar, and dragging inside it should pan
x = np.linspace(0, 10, 300)
fig, ax = plt.subplots()
line, = ax.plot(x, np.sin(x))
ax.set_title('Drag to pan, use the toolbar to zoom')
ax.grid(True)
print('✓', 'figure created (check that it appeared with a toolbar)')
```

```python
# 5. A slider: moving it should change the curve above without opening a new figure
def update(frequency=FloatSlider(min=0.5, max=5, step=0.5, value=1)):
    line.set_ydata(np.sin(frequency*x))
    fig.canvas.draw_idle()
interact(update);
```

If every line above has a check mark, the figure has a toolbar, and the slider moves the curve, the site works on this computer and browser.
