# Start here

This website is a complete Jupyter environment running in your web browser. The Python that executes these notebooks was downloaded along with the page and runs on your own computer. There is no server behind it, and you installed nothing. [JupyterLite](https://jupyterlite.readthedocs.io/) makes this possible, using [Pyodide](https://pyodide.org/), a build of Python for web browsers.

The site is the live example for *Build a Jupyter Site for Your Course*, a contribution to the [Faculty AI Exchange](https://facultyaiexchange.org/) describing how an AI coding assistant can move an instructor's existing notebooks onto a site like this one. The source for this site, including the configuration files, is at [github.com/william-pfalzgraff/jupyterlite-course-demo](https://github.com/william-pfalzgraff/jupyterlite-course-demo).

## What to try

Open each notebook and choose **Run > Run All Cells**. The first cell of each notebook installs what it needs, which takes a few seconds the first time.

2. **Carbon cycle**: a small climate model with sliders that change the emissions scenario and redraw the results.

Each notebook is adapted from published teaching material; the reference is at the top of each.

## Things worth knowing

- Anything you change is saved by your browser, on this computer, for this site only. Nobody else sees it, and the published notebooks are untouched. To start over, delete a notebook in the file browser and it comes back fresh.
- A cell that complains about a missing package means the first cell didn't run. Choose **Kernel > Restart Kernel and Run All Cells**.
- Python in a browser is slower than Python on a laptop, but not by much for work at this scale.
